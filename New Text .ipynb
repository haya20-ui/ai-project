```python
# Initialize parameters
w = 10.0  # Starting point
learning_rate = 0.1
epochs = 50
history = [] # To store values for plotting

# Training loop
for i in range(epochs):
    gradient = 2 * w
    w = w - learning_rate * gradient
    history.append(w)
    print(f"Epoch {i+1}: w = {w:.4f}")
```

```python
import numpy as np# 2

def run_gradient_descent(learning_rate, w_start=10.0, epochs=50):
    """
    Runs the Gradient Descent algorithm for a simple quadratic function.

    Args:
        learning_rate (float): The learning rate (alpha) for the optimizer.
        w_start (float): The initial value for the weight 'w'.
        epochs (int): The number of iterations to run.

    Returns:
        list: A list containing the history of 'w' values at each epoch.
    """
    w = w_start
    w_history = []
    
    print(f"\n--- Running Gradient Descent with learning_rate = {learning_rate} ---")
    
    for i in range(epochs):
        gradient = 2 * w  # The derivative of f(w) = w^2 is 2w
        w = w - learning_rate * gradient  # The GD update rule
        w_history.append(w)
        
        # Optional: Print progress for the first few epochs
        if i < 5 or i > epochs - 6:
            print(f"Epoch {i+1:2d}: w = {w:.6f}")

    print("--------------------------------------------------")
    return w_history

# --- Task 2: Experimenting with the Learning Rate (alpha) ---3

# 1. Run with a learning rate that is too small
history_small_lr = run_gradient_descent(learning_rate=0.01)

# 2. Run with a learning rate that is "just right"
history_good_lr = run_gradient_descent(learning_rate=0.1)

# 3. Run with a learning rate that is too large
history_large_lr = run_gradient_descent(learning_rate=0.8)

# The histories are now stored and ready for plotting as described in Slide 8.
print("\nFinal weight for small learning rate (0.01):", history_small_lr[-1])
print("Final weight for good learning rate (0.1):", history_good_lr[-1])
print("Final weight for large learning rate (0.8):", history_large_lr[-1])

```
